El servicio de venta de autos usados Rusty Bargain está desarrollando una aplicación para atraer nuevos clientes. Gracias a esa app, puedes averiguar rápidamente el valor de mercado de tu coche. Tienes acceso al historial: especificaciones técnicas, versiones de equipamiento y precios. Tienes que crear un modelo que determine el valor de mercado.
A Rusty Bargain le interesa:
- la calidad de la predicción;
- la velocidad de la predicción;
- el tiempo requerido para el entrenamiento

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el dataset
try:
    df = pd.read_csv('/datasets/car_data.csv')
except:
    df = pd.read_csv('car_data.csv') # Por si lo corres localmente

# Vista inicial
display(df.head())
df.info()
```

```python

df.columns = df.columns.str.lower()

# Renombrar manualmente algunas para que sean más claras (snake_case)
df = df.rename(columns={
    'vehicletype': 'vehicle_type',
    'registrationyear': 'registration_year',
    'registrationmonth': 'registration_month',
    'fueltype': 'fuel_type',
    'notrepaired': 'not_repaired',
    'datecreated': 'date_created',
    'numberofpictures': 'number_of_pictures',
    'postalcode': 'postal_code',
    'lastseen': 'last_seen',
    'datecrawled': 'date_crawled'
})


print(df.columns)
```

#### Primero se reenombraron las columnas para poder estandarizar y poder entender mejor la informacion

```python
print("Valores únicos en fotos:", df['number_of_pictures'].unique())
```

##### se verifica que en la columna "Numero de fotos" sean todas "0"

```python
cols_to_drop = ['number_of_pictures', 'date_crawled', 'date_created', 'last_seen', 'postal_code']
df = df.drop(columns=cols_to_drop, axis=1)

print("Columnas restantes:", df.columns)
```

##### Se eliminan las columnas irrelevantes

```python
cols_con_nulos = ['vehicle_type', 'gearbox', 'model', 'fuel_type', 'not_repaired']

for col in cols_con_nulos:
    df[col] = df[col].fillna('unknown')

print(df.isna().sum())
```

##### Se eliminan los valores nulos con "unknow" ya que como son categoricas nos permite usar la informacion sin inventarla, ademas en la columna de "not_repaired" no tenemos la seguridad si el auto esta reparado o no

```python
df = df[df['price'] > 100]
df = df[(df['power'] > 40) & (df['power'] < 1000)]
df = df[(df['registration_year'] >= 1960) & (df['registration_year'] <= 2016)]

print(f"Datos después de limpieza: {df.shape}")
df.describe()
```

##### Se eliminaron las filas que tienen anomalias en sus datos
* Precios absurdos como 0 o 1
* Caballos de fuerza muy bajos como 0 o muy exagerados como mas de 2000
* coches con año de fabriacion irreales como 1000 o 9000

## Entrenamiento del modelo 

```python
from sklearn.metrics import mean_squared_error

def rmse(y_true, y_pred):
    return mean_squared_error(y_true, y_pred) ** 0.5
```

```python
features = df.drop('price', axis=1)
target = df['price']
```

##### Preparacion de versiones de datos

```python
from sklearn.preprocessing import OrdinalEncoder


features_lgbm = features.copy()

cat_columns = features_lgbm.select_dtypes(include='object').columns
for col in cat_columns:
    features_lgbm[col] = features_lgbm[col].astype('category')


features_oe = features.copy()
encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
features_oe[cat_columns] = encoder.fit_transform(features_oe[cat_columns])


features_ohe = pd.get_dummies(features, drop_first=True)

print("Shape features original/LGBM:", features_lgbm.shape)
print("Shape features Ordinal:", features_oe.shape)
print("Shape features OHE:", features_ohe.shape) 
```

##### Division de datos

```python
from sklearn.model_selection import train_test_split

def split_data(features, target):
    features_train, features_temp, target_train, target_temp = train_test_split(
        features, target, test_size=0.4, random_state=12345
    )
    features_valid, features_test, target_valid, target_test = train_test_split(
        features_temp, target_temp, test_size=0.5, random_state=12345
    )
    return features_train, features_valid, features_test, target_train, target_valid, target_test

X_train_ohe, X_valid_ohe, X_test_ohe, y_train, y_valid, y_test = split_data(features_ohe, target)

X_train_oe, X_valid_oe, X_test_oe, _, _, _ = split_data(features_oe, target)


X_train_lgbm, X_valid_lgbm, X_test_lgbm, _, _, _ = split_data(features_lgbm, target)

print("Tamaño del set de entrenamiento:", X_train_ohe.shape)
print("Tamaño del set de validación:", X_valid_ohe.shape)
print("Tamaño del set de prueba:", X_test_ohe.shape)
```

##### Modelo de regresion lineal

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import time

model_lr = LinearRegression()

start_train = time.time()
model_lr.fit(X_train_ohe, y_train)
end_train = time.time()
time_lr_train = end_train - start_train

start_pred = time.time()
preds_lr = model_lr.predict(X_valid_ohe)
end_pred = time.time()
time_lr_pred = end_pred - start_pred

rmse_lr = mean_squared_error(y_valid, preds_lr, squared=False)

print("Resultados Regresión Lineal:")
print(f"Tiempo de entrenamiento: {time_lr_train:.4f} s")
print(f"Tiempo de predicción:    {time_lr_pred:.4f} s")
print(f"RMSE (Validación):       {rmse_lr:.2f} euros")
```

##### Modelo bosque aleatorio

```python
from sklearn.ensemble import RandomForestRegressor

best_rmse_rf = float('inf')
best_depth = 0
best_est = 0

for depth in [5, 10, 15]:
    for est in [10, 50]:
        model_rf = RandomForestRegressor(n_estimators=est, max_depth=depth, random_state=12345, n_jobs=-1)
        
        model_rf.fit(X_train_oe, y_train)
        
        preds_rf = model_rf.predict(X_valid_oe)
        
        rmse = mean_squared_error(y_valid, preds_rf, squared=False)
        print(f"Depth: {depth}, Estimators: {est} -> RMSE: {rmse:.2f}")
        
        if rmse < best_rmse_rf:
            best_rmse_rf = rmse
            best_depth = depth
            best_est = est
            best_model_rf = model_rf

print("-" * 30)
print(f"Mejor Random Forest: Depth={best_depth}, Est={best_est}")
print(f"Mejor RMSE RF: {best_rmse_rf:.2f}")
```

#####  LightGBM 

```python
import lightgbm as lgb
import time

model_lgbm = lgb.LGBMRegressor(random_state=12345)

start_train = time.time()
model_lgbm.fit(X_train_lgbm, y_train)
end_train = time.time()
time_lgbm_train = end_train - start_train

start_pred = time.time()
preds_lgbm = model_lgbm.predict(X_valid_lgbm)
end_pred = time.time()
time_lgbm_pred = end_pred - start_pred

rmse_lgbm = mean_squared_error(y_valid, preds_lgbm, squared=False)

print(f"LightGBM (Base) - Train: {time_lgbm_train:.4f}s | Pred: {time_lgbm_pred:.4f}s | RMSE: {rmse_lgbm:.2f}")

model_lgbm_tuned = lgb.LGBMRegressor(n_estimators=500, learning_rate=0.1, max_depth=10, random_state=12345)

start_train = time.time()
model_lgbm_tuned.fit(X_train_lgbm, y_train)
end_train = time.time()
time_lgbm_tuned_train = end_train - start_train

start_pred = time.time()
preds_lgbm_tuned = model_lgbm_tuned.predict(X_valid_lgbm)
end_pred = time.time()
time_lgbm_tuned_pred = end_pred - start_pred

rmse_lgbm_tuned = mean_squared_error(y_valid, preds_lgbm_tuned, squared=False)

print(f"LightGBM (Tuned) - Train: {time_lgbm_tuned_train:.4f}s | Pred: {time_lgbm_tuned_pred:.4f}s | RMSE: {rmse_lgbm_tuned:.2f}")
```

##### CatBoost

```python
from catboost import CatBoostRegressor


cat_features = list(X_train_lgbm.select_dtypes(include=['category']).columns)

model_cb = CatBoostRegressor(loss_function='RMSE', iterations=200, learning_rate=0.1, verbose=50, random_seed=12345)

start_train = time.time()
model_cb.fit(X_train_lgbm, y_train, cat_features=cat_features)
end_train = time.time()
time_cb_train = end_train - start_train

start_pred = time.time()
preds_cb = model_cb.predict(X_valid_lgbm)
end_pred = time.time()
time_cb_pred = end_pred - start_pred

rmse_cb = mean_squared_error(y_valid, preds_cb, squared=False)

print("-" * 30)
print(f"CatBoost - Train: {time_cb_train:.4f}s | Pred: {time_cb_pred:.4f}s | RMSE: {rmse_cb:.2f}")
```

##### XGBoost

```python
import xgboost as xgb

model_xgb = xgb.XGBRegressor(objective='reg:squarederror', n_estimators=100, random_state=12345)

start_train = time.time()
model_xgb.fit(X_train_ohe, y_train)
end_train = time.time()
time_xgb_train = end_train - start_train

start_pred = time.time()
preds_xgb = model_xgb.predict(X_valid_ohe)
end_pred = time.time()
time_xgb_pred = end_pred - start_pred

rmse_xgb = mean_squared_error(y_valid, preds_xgb, squared=False)

print("-" * 30)
print(f"XGBoost - Train: {time_xgb_train:.4f}s | Pred: {time_xgb_pred:.4f}s | RMSE: {rmse_xgb:.2f}")
```

## Análisis del modelo

* LightGBM (Tuned) no solo tiene el mejor error (RMSE 1501), sino que entrenó en 6 segundos.
* XGBoost tuvo un error decente (1589), pero tardó 135 segundos (más de 2 minutos). ¡LightGBM fue 20 veces más rápido!
* CatBoost fue el más rápido prediciendo (0.08s), pero su error fue un poco más alto (quizás le faltaron más iteraciones, pero para este ejercicio está bien).

```python
resultados = {
    'Modelo': ['Regresión Lineal', 'Random Forest', 'LightGBM Base', 'LightGBM Tuned', 'CatBoost', 'XGBoost'],
    'RMSE (Validación)': [2524.64, 1599.19, 1572.28, 1501.20, 1645.66, 1589.64],
    'Tiempo Entrenamiento (s)': [6.30, 60.0, 2.05, 6.20, 28.74, 135.48], 
    'Tiempo Predicción (s)': [0.17, 0.10, 0.29, 0.90, 0.08, 0.30]
}

df_resultados = pd.DataFrame(resultados)


df_resultados = df_resultados.sort_values(by='RMSE (Validación)', ascending=True)

display(df_resultados)
```

 **Análisis de Modelos:**
 1.  **Calidad (RMSE):** El modelo **LightGBM Tuned** obtuvo el mejor desempeño con un error de **1501 euros**, superando significativamente a la Regresión Lineal (2524 euros) y superando por poco a XGBoost (1589 euros) y Random Forest (1599 euros).

2.  **Velocidad de Entrenamiento:** Aquí es donde **LightGBM** brilla. El modelo base entrenó en solo **2 segundos** y el ajustado en **6 segundos**. Comparado con XGBoost que tardó **135 segundos**, LightGBM es indiscutiblemente la opción más eficiente para re-entrenamientos frecuentes.

3.  **Velocidad de Predicción:** Aunque **CatBoost** fue el más rápido en predecir (0.08s), **LightGBM** se mantiene en un rango muy aceptable (menos de 1 segundo). Para una App de usuario, cualquier tiempo menor a 1 segundo es imperceptible.
