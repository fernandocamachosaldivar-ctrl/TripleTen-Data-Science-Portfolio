# Informe de Solución: Predicción de Churn - Interconnect

## 1. Pasos del plan realizado
Se siguieron todos los pasos propuestos en el plan de trabajo inicial, cumpliendo con el ciclo de vida completo del proyecto:

1.  **Análisis Exploratorio y Limpieza:** Se unificaron las cuatro fuentes de datos (contract, personal, internet, phone). Se corrigieron tipos de datos, específicamente en `TotalCharges
  
2.  **Ingeniería de Características:** Se creó la variable objetivo (target) y se calculó la antigüedad del cliente (tenure_days) basándose en la fecha de corte del 1 de febrero de 2020. Se aplicó One-Hot Encoding y escalamiento de variables.

5.  **Entrenamiento y Selección:** Se probaron tres modelos: Regresión Logística (baseline), Random Forest y LightGBM.

6.  **Optimización:** Se realizó una búsqueda de hiperparámetros mediante GridSearchCV para el modelo final.

7.  **Evaluación:** Se validaron los resultados en un conjunto de prueba (test set) que el modelo nunca había visto.

**Pasos omitidos:** No se omitió ningún paso principal. Sin embargo, se decidió no profundizar en modelos de Redes Neuronales ya que los modelos de Gradient Boosting alcanzaron métricas superiores al objetivo de 0.88 de forma más eficiente.

## 2. Dificultades encontradas y soluciones
*   **Datos ausentes tras la unión:** Al unir los archivos, surgieron valores nulos en servicios adicionales. Solución: Se determinó que estos nulos representaban clientes que no contrataron dichos servicios, por lo que se imputaron con la etiqueta "No", preservando la integridad de la muestra.
*   **Valores vacíos en TotalCharges:** Se detectaron espacios en blanco en esta columna que impedían su conversión a numérica. Solución: Se identificó que correspondían a clientes nuevos con antigüedad cero. Se reemplazaron por 0 para evitar la pérdida de estos registros.
*   **Desbalance de clases:** Había significativamente más clientes activos que los que habían cancelado. Solución: Se utilizó el parámetro stratify al dividir los datos y se priorizó la métrica AUC-ROC, que es robusta frente a desbalances.
*   **Advertencias de memoria (SettingWithCopyWarning):** Al escalar datos. Solución: Se utilizó el método .copy() para asegurar que las transformaciones se realizaran sobre copias independientes de los DataFrames.

## 3. Pasos clave para resolver la tarea
*   **Cálculo de tenure_days:** Transformar las fechas de inicio y fin en una variable de "días de permanencia" fue el paso más crítico. La importancia de características demostró que esta es la variable con mayor poder predictivo.
*   **Tratamiento de nulos con lógica de negocio:** En lugar de usar métodos estadísticos (media/mediana), usar el contexto del negocio para llenar nulos con "No" permitió mantener la realidad operativa de la empresa.
*   **Uso de LightGBM:** La elección de un modelo de Gradient Boosting permitió capturar relaciones no lineales complejas entre los cargos mensuales y el tipo de internet, lo que disparó la calidad del modelo por encima de los métodos lineales.

## 4. Modelo final y nivel de calidad
El modelo final seleccionado es **LightGBM Classifier**, optimizado con los siguientes hiperparámetros:
*   learning_rate: 0.1
*   max_depth: 3
*   n_estimators: 500
*   num_leaves: 21

### Nivel de Calidad Alcanzado:
*   **AUC-ROC (Test set): 0.9001**
*   **Exactitud (Accuracy): 0.8630**

**Conclusión técnica:** El modelo demuestra una capacidad excepcional (0.90) para distinguir entre un cliente que cancelará y uno que no. Al superar el umbral de 0.88, se garantiza el cumplimiento de los requerimientos de Interconnect para la asignación de presupuestos de retención y marketing preventivo.
